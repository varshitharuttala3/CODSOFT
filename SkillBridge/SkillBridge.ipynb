```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = {
    "Title": [
        "AI/ML Internship",
        "Web Development Internship",
        "Data Science Internship",
        "Cloud Computing Internship",
        "Cybersecurity Internship",
        "Deep Learning Course",
        "Full Stack Development Course",
        "AWS Cloud Course",
        "Python for Data Science Course",
        "Ethical Hacking Course"
    ],

    "Skills": [
        "python machine learning deep learning tensorflow",
        "html css javascript react",
        "python data analysis pandas numpy statistics",
        "aws cloud devops docker kubernetes",
        "network security penetration testing linux",
        "neural networks cnn rnn pytorch",
        "frontend backend node react mongodb",
        "aws ec2 s3 cloud architecture",
        "python pandas numpy visualization",
        "cybersecurity hacking ethical security"
    ],

    "Type": [
        "Internship",
        "Internship",
        "Internship",
        "Internship",
        "Internship",
        "Course",
        "Course",
        "Course",
        "Course",
        "Course"
    ]
}

df = pd.DataFrame(data)
df.head()
```

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df["Skills"])
```

```python
def recommend(user_skills):

    user_vector = tfidf.transform([user_skills])
    similarity = cosine_similarity(user_vector, tfidf_matrix)

    scores = similarity[0]

    top_indices = scores.argsort()[-5:][::-1]

    recommendations = df.iloc[top_indices]

    return recommendations[["Title", "Type"]]
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

st.set_page_config(page_title="SkillBridge", layout="centered")

st.title("🎓 SkillBridge")
st.subheader("AI Internship & Course Recommender")

# Sample Data
data = {
    "Title": [
        "AI/ML Internship",
        "Web Development Internship",
        "Data Science Internship",
        "Cloud Computing Internship",
        "Cybersecurity Internship",
        "Deep Learning Course",
        "Full Stack Development Course",
        "AWS Cloud Course",
        "Python for Data Science Course",
        "Ethical Hacking Course"
    ],
    "Skills": [
        "python machine learning deep learning tensorflow",
        "html css javascript react",
        "python data analysis pandas numpy statistics",
        "aws cloud devops docker kubernetes",
        "network security penetration testing linux",
        "neural networks cnn rnn pytorch",
        "frontend backend node react mongodb",
        "aws ec2 s3 cloud architecture",
        "python pandas numpy visualization",
        "cybersecurity hacking ethical security"
    ],
    "Type": [
        "Internship","Internship","Internship","Internship","Internship",
        "Course","Course","Course","Course","Course"
    ]
}

df = pd.DataFrame(data)

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df["Skills"])

def recommend(user_skills):
    user_vector = tfidf.transform([user_skills])
    similarity = cosine_similarity(user_vector, tfidf_matrix)
    scores = similarity[0]
    top_indices = scores.argsort()[-5:][::-1]
    results = df.iloc[top_indices].copy()
    results["Match Score (%)"] = (scores[top_indices] * 100).round(2)
    return results[["Title","Type","Match Score (%)"]]

skills = st.text_input("Enter your skills (e.g. python machine learning)")

if st.button("Get Recommendations"):
    if skills:
        results = recommend(skills)
        st.dataframe(results)
    else:
        st.warning("Please enter your skills.")
```
